```python
import numpy as np
```

```python
temp=np.random.randint(18,40,168)
re_shape=np.reshape(temp,(7,24))
```

```python
avg_day=np.mean(re_shape,axis=1)
max_day=np.max(re_shape,axis=1)
min_day=np.min(re_shape,axis=1)
hot_hour=temp[temp>35]
average=np.average(temp)
min=np.min(temp)
max=np.max(temp)
```

```python

print("------weekly temperature analyzer------ \n")
print()
print("__temperature of each day__ \n")
print(re_shape)
print()
for i in range(7):
   print("average temp in day ",i+1," : ",avg_day[i])
   print("max temp in day ",i+1," : ",max_day[i])
   print("min temp in day ",i+1," : ",min_day[i])
print()
print("hot hour : ",hot_hour)
print()
print("average temp in week : ",average)
print()
print("min temp in week : ",min)
print()
print("max temp in week  : ",max)
```
